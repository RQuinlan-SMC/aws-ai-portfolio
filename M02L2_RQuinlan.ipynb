```python
import pandas as pd
df = pd.read_csv('m2_metric_scenarios.csv')

df['my_accuracy']  = (df['true_positive']+df['true_negative'])/(df['true_negative'] + df['true_positive'] + df['false_negative'] + df['false_positive'] )
df['my_precision'] = (df['true_positive']/(df['true_positive'] + df['false_positive']))
df['my_recall']    = (df['true_positive']/(df['true_positive'] + df['false_negative']))
df['my_f1']        = (2*(df['my_precision']*df['my_recall'])/(df['my_precision']+df['my_recall']))

df['total_error_cost'] = ((df['false_positive']*df['dollar_cost_per_false_positive']) + (df['false_negative'] * df['dollar_cost_per_false_negative']))

print(df[['scenario_id','accuracy','my_accuracy', 'precision', 'my_precision','recall','my_recall','f1','my_f1','total_error_cost']])
print('------------------------------------------------------------------')
print('------------------------------------------------------------------')
# check your work against the columns already in the file
print(df[['scenario_id','accuracy','my_accuracy', 'precision', 'my_precision','recall','my_recall','f1','my_f1','total_error_cost']].sort_values('total_error_cost',ascending=False))
print('------------------------------------------------------------------')
print('------------------------------------------------------------------')
print('S4 Sweep')
print('------------------------------------------------------------------')
# a simple sweep. At each step you accept 4 more false positives
# in exchange for 1 fewer false negative.
fp, fn = 130, 95
cost_fp, cost_fn = 300, 3200
for step in range(0, 6):
    f_p = fp + step
    f_n = fn - step
    total = (f_p*cost_fp) + (f_n * cost_fn)
    print(step, f_p, f_n, total)
print('------------------------------------------------------------------')
```
